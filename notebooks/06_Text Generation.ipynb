## Text Generation Pipeline

```
Prompt

↓

Tokenizer

↓

Input IDs

↓

Transformer

↓

Next Token Prediction

↓

Generated Text
```

The `generate()` function repeatedly predicts the next token until the stopping condition is reached.

````python
from transformers import AutoTokenizer, AutoModelForCausalLM

model_name = "gpt2"

tokenizer = AutoTokenizer.from_pretrained(model_name)

model = AutoModelForCausalLM.from_pretrained(model_name)
````

````python
prompt = "Artificial Intelligence is"

inputs = tokenizer(
    prompt,
    return_tensors="pt"
)

print(inputs)
````

````python
outputs = model.generate(
    **inputs,
    max_new_tokens=30
)

generated_text = tokenizer.decode(
    outputs[0],
    skip_special_tokens=True
)

print(generated_text)
````

## Behind `generate()`

```
Input IDs

↓

Transformer

↓

Logits

↓

Softmax

↓

Next Token

↓

Append Token

↓

Repeat
```

This loop continues until an end-of-sequence token is generated or the maximum number of new tokens is reached.

````python
for tokens in [10, 30, 60]:
    
    output = model.generate(
        **inputs,
        max_new_tokens=tokens
    )

    print("="*80)
    print(f"max_new_tokens = {tokens}\n")

    print(tokenizer.decode(output[0], skip_special_tokens=True))
````

## `max_new_tokens`

```
Small Value

↓

Short Response

────────────

Large Value

↓

Long Response
```

This parameter controls how many **new tokens** the model is allowed to generate.

````python
temperatures = [0.2,0.7,1.0,1.5]

for temp in temperatures:

    output = model.generate(
        **inputs,
        max_new_tokens=30,
        temperature=temp,
        do_sample=True
    )

    print("="*80)
    print(f"Temperature = {temp}\n")

    print(tokenizer.decode(output[0], skip_special_tokens=True))
````

## Temperature

```
Low Temperature

↓

More Predictable

────────────

High Temperature

↓

More Creative
```

Temperature controls how random the next-token selection becomes.

````python
for k in [10,50,100]:

    output = model.generate(
        **inputs,
        max_new_tokens=30,
        do_sample=True,
        top_k=k
    )

    print("="*80)
    print(f"top_k = {k}\n")

    print(tokenizer.decode(output[0], skip_special_tokens=True))
````

## Top-k Sampling

```
Vocabulary

↓

Keep Top K Tokens

↓

Randomly Select One
```

The model samples only from the K most probable tokens.

````python
for p in [0.5,0.8,0.95]:

    output = model.generate(
        **inputs,
        max_new_tokens=30,
        do_sample=True,
        top_p=p
    )

    print("="*80)
    print(f"top_p = {p}\n")

    print(tokenizer.decode(output[0], skip_special_tokens=True))
````

## Top-p (Nucleus Sampling)

```
Vocabulary

↓

Keep Tokens Until

Probability ≥ p

↓

Sample
```

Instead of selecting a fixed number of tokens, Top-p selects the smallest set whose cumulative probability exceeds `p`.

````python
greedy = model.generate(
    **inputs,
    max_new_tokens=30,
    do_sample=False
)

sampled = model.generate(
    **inputs,
    max_new_tokens=30,
    do_sample=True
)

print("Greedy\n")
print(tokenizer.decode(greedy[0],skip_special_tokens=True))

print("\n")

print("Sampling\n")
print(tokenizer.decode(sampled[0],skip_special_tokens=True))
````

## Greedy vs Sampling

```
Highest Probability

↓

Greedy Search

────────────

Probability Distribution

↓

Random Sampling
```

Greedy decoding always selects the most probable token.

Sampling introduces controlled randomness.

````python
output = model.generate(
    **inputs,
    max_new_tokens=40,
    repetition_penalty=1.8
)

print(tokenizer.decode(output[0],skip_special_tokens=True))
````

## Repetition Penalty

This parameter discourages the model from repeatedly generating the same words or phrases.

Higher values reduce repetition but may also reduce fluency.

````python
beam = model.generate(
    **inputs,
    max_new_tokens=30,
    num_beams=5,
    early_stopping=True
)

print(tokenizer.decode(beam[0],skip_special_tokens=True))
````

## Beam Search

```
Sentence

↓

Beam 1

Beam 2

Beam 3

↓

Best Sequence
```

Instead of following a single path, Beam Search keeps multiple candidate sequences and selects the most likely one.
