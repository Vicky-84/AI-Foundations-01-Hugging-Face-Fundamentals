# 🔤 Notebook 03 - Understanding Tokenizers

In this notebook, you'll learn how tokenizers convert human-readable text into numbers that Transformer models can understand.

By the end of this notebook, you'll be able to:

- Load a tokenizer
- Encode text
- Decode tokens
- Understand Token IDs
- Understand Attention Masks

## Tokenization Workflow

```
Text

↓

Tokenizer

↓

Tokens

↓

Token IDs

↓

Transformer
```

A Transformer model cannot understand text directly.

The tokenizer acts as a translator between human language and the model.

````python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("gpt2")
````

````python
text = "Artificial Intelligence is amazing."

tokens = tokenizer.tokenize(text)

print(tokens)
````

### Expected Output

```python
['Art', 'ificial', 'ĠIntelligence', 'Ġis', 'Ġamazing', '.']
```

> The exact tokens may differ depending on the tokenizer and model.

## What Happened?

```
Artificial Intelligence is amazing.

↓

Tokenizer

↓

['Art', 'ificial', 'ĠIntelligence', 'Ġis', 'Ġamazing', '.']
```

Instead of processing whole words, GPT-2 breaks text into **subword tokens**.

This helps the model handle unknown words more efficiently.

````python
token_ids = tokenizer.convert_tokens_to_ids(tokens)

print(token_ids)
````

### Expected Output

```python
[8001, 9542, 4430, 318, 4998, 13]
```

The numbers will vary depending on the tokenizer vocabulary.

````python
decoded = tokenizer.convert_ids_to_tokens(token_ids)

print(decoded)
````

## Tokenization Pipeline

```
Sentence

↓

Tokens

↓

Token IDs

↓

Transformer

↓

Prediction
```

The Transformer never sees words.

It only processes numerical IDs.

````python
encoding = tokenizer(
    text,
    return_tensors="pt"
)

print(encoding)
````

````python
print("Input IDs:")
print(encoding["input_ids"])

print()

print("Attention Mask:")
print(encoding["attention_mask"])
````

## Understanding the Output

```
Text

↓

Tokenizer

↓

Input IDs
Attention Mask
```

**Input IDs**

The numerical representation of each token.

**Attention Mask**

- 1 → Real token
- 0 → Padding token

````python
print(tokenizer.decode(encoding["input_ids"][0]))
````

## Decoding

```
Input IDs

↓

Tokenizer

↓

Original Text
```

Decoding converts token IDs back into readable text.

````python
sentence = "Machine Learning is fun."

encoded = tokenizer(
    sentence,
    return_tensors="pt"
)

print(encoded)
````

````python
print(encoded["input_ids"].shape)
````

## Tensor Shape

```
(batch_size, sequence_length)
```

Example:

```
torch.Size([1, 6])
```

- Batch Size = 1 sentence
- Sequence Length = 6 tokens

````python
print(tokenizer.vocab_size)
````

## Vocabulary Size

Every tokenizer has a fixed vocabulary.

For GPT-2:

```
Vocabulary

↓

50257 Tokens
```

Each token has a unique numerical ID.

````python
word = "cats"

ids = tokenizer.encode(word)

print(ids)

print(tokenizer.decode(ids))
````
