## Auto Classes

```
Model Name

↓

Auto Class

↓

Correct Architecture

↓

Ready to Use Model
```

Instead of manually importing every model architecture, Hugging Face automatically detects and loads the correct model based on the model name.

````python
from transformers import AutoModel
from transformers import AutoTokenizer

model_name = "bert-base-uncased"

tokenizer = AutoTokenizer.from_pretrained(model_name)

model = AutoModel.from_pretrained(model_name)
````

````python
print(type(model))
````

### Expected Output

```python
<class 'transformers.models.bert.modeling_bert.BertModel'>
```

Although we used `AutoModel`, Hugging Face actually loaded a `BertModel`.

## Behind the Scenes

```
bert-base-uncased

↓

Read config.json

↓

architecture = BertModel

↓

Load BertModel
```

The Auto Class checks the model's configuration and instantiates the correct architecture automatically.

````python
print(model.config)
````

````python
print("Hidden Size :", model.config.hidden_size)
print("Layers      :", model.config.num_hidden_layers)
print("Attention   :", model.config.num_attention_heads)
print("Vocabulary  :", model.config.vocab_size)
````

Example Output

```text
Hidden Size : 768
Layers      : 12
Attention   : 12
Vocabulary  : 30522
```

## What is Config?

Every pretrained model has a configuration file.

```
config.json

↓

Hidden Size

Layers

Attention Heads

Vocabulary Size

Dropout

Maximum Position Embeddings
```

The configuration defines the model architecture—not the learned weights.

````python
from transformers import AutoConfig

config = AutoConfig.from_pretrained("gpt2")

print(config)
````

## AutoConfig

```
Model Name

↓

Download config.json

↓

Python Config Object
```

Use `AutoConfig` when you want to inspect a model without downloading its weights.

````python
print(type(config))
````

### Expected Output

```python
<class 'transformers.models.gpt2.configuration_gpt2.GPT2Config'>
```

## Choosing the Right Auto Class

| Auto Class | Purpose |
|------------|---------|
| AutoTokenizer | Load tokenizer |
| AutoConfig | Load configuration |
| AutoModel | Base model (embeddings + encoder/decoder) |
| AutoModelForCausalLM | Text generation |
| AutoModelForSeq2SeqLM | Translation / Summarization |
| AutoModelForSequenceClassification | Classification |
| AutoModelForTokenClassification | NER |
| AutoModelForQuestionAnswering | QA |

````python
from transformers import AutoModelForCausalLM

gpt_model = AutoModelForCausalLM.from_pretrained("gpt2")

print(type(gpt_model))
````

### Expected Output

```python
<class 'transformers.models.gpt2.modeling_gpt2.GPT2LMHeadModel'>
```

Notice that Hugging Face selected a language-generation model instead of the base GPT-2 model.

## AutoModel vs AutoModelForCausalLM

```
AutoModel
│
├── Hidden States
│
└── Embeddings

AutoModelForCausalLM
│
├── Hidden States
├── LM Head
└── Next Token Prediction
```

Use **AutoModel** when you need embeddings or hidden states.

Use **AutoModelForCausalLM** when generating text.

````python
print(gpt_model.config.model_type)
print(gpt_model.config.vocab_size)
print(gpt_model.config.n_layer)
````
