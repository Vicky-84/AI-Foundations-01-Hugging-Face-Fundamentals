## Loading Different Models

One of the biggest advantages of Hugging Face is the unified API.

Only the model name changes.

```
GPT-2

↓

BERT

↓

T5

↓

Llama

↓

Gemma

↓

Mistral

↓

Same API
```

````python
models = [
    "gpt2",
    "distilgpt2",
    "bert-base-uncased",
    "t5-small"
]

for model_name in models:
    print(model_name)
````

## Popular Model Categories

````python
from transformers import pipeline

generator = pipeline(
    "text-generation",
    model="distilgpt2"
)

response = generator(
    "Machine Learning is",
    max_new_tokens=30
)

print(response[0]["generated_text"])
````

````python
print(model.config)
````

## What is `config`?

Every Hugging Face model comes with a configuration object.

It stores important information such as:

- Hidden Size
- Number of Layers
- Number of Attention Heads
- Vocabulary Size
- Maximum Position Embeddings

We'll revisit these parameters when building a Transformer from scratch.

````python
print("Hidden Size :", model.config.hidden_size)
print("Layers      :", model.config.num_hidden_layers)
print("Attention   :", model.config.num_attention_heads)
print("Vocabulary  :", model.config.vocab_size)
````

Example Output

```

Attention   : 12
Vocabulary  : 30522
```
