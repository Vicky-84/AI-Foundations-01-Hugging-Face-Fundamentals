## What is a Pipeline?

```
Task

↓

Pipeline

↓

Tokenizer

↓

Model

↓

Prediction
```

A Pipeline is a high-level API that automatically loads the correct tokenizer and model for a specific task.

````python
from transformers import pipeline
````

````python
generator = pipeline(
    "text-generation",
    model="gpt2"
)

print(generator)
````

### Expected Output

```
TextGenerationPipeline(...)
```

The pipeline has automatically loaded:

- Tokenizer
- Model
- Generation utilities

````python
result = generator(
    "Artificial Intelligence is",
    max_new_tokens=40
)

print(result)
````

````python
print(result[0]["generated_text"])
````

## Pipeline Internals

```
Pipeline
│
├── Tokenizer
├── Model
└── Task Logic
```

A Pipeline is simply a wrapper around the tokenizer and model with task-specific preprocessing and postprocessing.

````python
sentiment = pipeline(
    "sentiment-analysis"
)

result = sentiment(
    "I absolutely love learning Artificial Intelligence!"
)

print(result)
````

### Expected Output

```python
[
 {'label': 'POSITIVE', 'score': 0.999}
]
```

````python
classifier = pipeline(
    "zero-shot-classification"
)

candidate_labels = [
    "technology",
    "sports",
    "finance"
]

result = classifier(
    "Transformers are changing AI.",
    candidate_labels
)

print(result)
````

## Zero-Shot Classification

```
Sentence

↓

Candidate Labels

↓

Most Likely Label
```

The model predicts the most relevant label without requiring additional training.

## Multiple NLP Tasks

```
                Pipeline

       ┌────────┼────────┐

       ▼        ▼        ▼

Text Gen   Sentiment   QA

       ▼        ▼        ▼

Translation  Summary  Classification
```

Changing the task changes the model behavior while keeping the API almost identical.

````python
from transformers import pipeline

generator = pipeline(
    "text-generation",
    model="gpt2"
)

print(generator("Python is", max_new_tokens=20)[0]["generated_text"])

print("-" * 80)

print(generator("Artificial Intelligence", max_new_tokens=20)[0]["generated_text"])
````

````python
generator = pipeline(
    "text-generation",
    model="distilgpt2"
)

print(generator("Python is", max_new_tokens=20)[0]["generated_text"])
````

## GPT-2 vs DistilGPT-2

```
Prompt

↓

GPT-2

↓

Generated Text

──────────────

Prompt

↓

DistilGPT-2

↓

Generated Text
```

Both models perform text generation, but DistilGPT-2 is a smaller, faster version of GPT-2. Compare the generated outputs and observe any differences in style or coherence.

````python
prompts = [
    "The future of AI",
    "Machine Learning",
    "My favourite programming language"
]

for prompt in prompts:
    print("=" * 80)
    print(f"Prompt: {prompt}\n")
    output = generator(prompt, max_new_tokens=25)
    print(output[0]["generated_text"])
    print()
````
