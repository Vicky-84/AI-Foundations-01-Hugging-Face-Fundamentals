## Hugging Face Workflow

```
              User Prompt
                   │
                   ▼
             Hugging Face
                   │
      ┌────────────┴────────────┐
      │                         │
Tokenizer                 Pretrained Model
      │                         │
      └────────────┬────────────┘
                   ▼
           Generated Response
```

For now, treat Hugging Face as a toolbox that provides pretrained models and tokenizers through a unified API.

````python
# Install required libraries (Colab)

!pip install -q transformers datasets sentencepiece accelerate
````

````python
from transformers import AutoTokenizer
from transformers import AutoModelForCausalLM

model_name = "gpt2"

tokenizer = AutoTokenizer.from_pretrained(model_name)

model = AutoModelForCausalLM.from_pretrained(model_name)

print("✅ Model Loaded Successfully")
````

### Expected Output

```
Model Loaded Successfully
```

> **Note:** The first execution may take a minute because the model is downloaded from the Hugging Face Hub. Subsequent runs use the cached version.

````python
prompt = "Artificial Intelligence is"

inputs = tokenizer(
    prompt,
    return_tensors="pt"
)

print(inputs)
````

## What Happened?

```
Prompt

↓

Tokenizer

↓

Input IDs + Attention Mask
```

The tokenizer converts human-readable text into numerical tensors that the Transformer model can understand.

````python
print(inputs["input_ids"])
print(inputs["attention_mask"])
````

````python
outputs = model.generate(
    **inputs,
    max_new_tokens=30
)

print(outputs)
````

## Generation Pipeline

```
Input IDs

↓

Transformer

↓

Logits

↓

Softmax

↓

Next Token

↓

Repeat

↓

Generated Sequence
```

The `generate()` method performs this loop repeatedly until the requested number of tokens is generated or an end-of-sequence token is produced.

````python
generated_text = tokenizer.decode(
    outputs[0],
    skip_special_tokens=True
)

print(generated_text)
````

### Expected Output

```
Artificial Intelligence is becoming an essential part of...
```

Your output will differ because text generation is probabilistic.
